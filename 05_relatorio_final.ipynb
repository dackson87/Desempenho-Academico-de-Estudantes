## RELATÓRIO FINAL — Projeto de Machine Learning

Predição da Nota Final de Estudantes com Base em Hábitos de Estudo e Fatores Socioeducacionais

**1. Resumo Executivo**

Este projeto teve como objetivo construir um modelo de Machine Learning capaz de prever a nota final de estudantes utilizando um conjunto de variáveis relacionadas a hábitos de estudo, condições familiares, participação em atividades extracurriculares e aspectos de saúde.

Foram realizadas análises exploratórias, tratamento completo do dataset, criação de novas variáveis, testes com múltiplos modelos e otimização via RandomizedSearch. O modelo final obteve excelente desempenho, com:

- R² (teste): 0.9645

- MSE: 0.0361

- RMSE: 0.1899

- MAE: 0.0987

Esses resultados indicam que o modelo consegue prever com alta precisão a nota final do aluno, explicando mais de 96% da variância do target.

O projeto foi documentado, organizado em um repositório GitHub e finalizado com relatório técnico e apresentação.

**2. Introdução**

**2.1 Contexto e Problema de Negócio**

A predição de desempenho acadêmico é uma das aplicações mais relevantes de Machine Learning na educação. Antecipar o rendimento de estudantes permite intervenções mais rápidas, direcionamento de reforço escolar e elaboração de políticas educacionais.

Este projeto trabalha com um dataset de 2.510 estudantes contendo informações pessoais, hábitos e condições familiares, com o objetivo de prever a nota final do aluno.

**2.2 Objetivo Geral**

Construir um pipeline completo de Machine Learning capaz de:

- analisar o dataset (EDA),

- preprocessar dados,

- treinar e comparar modelos,

- otimizar hiperparâmetros,

- selecionar o melhor modelo para predição da nota final.

**2.3 Metodologia Utilizada**

A metodologia seguiu as etapas:

1. Exploração dos Dados (EDA)

2. Pré-processamento

3. Modelagem

4. Otimização

5. Conclusões e documentação final

Todas as etapas foram feitas em Python (Pandas, Scikit-Learn, Matplotlib, Seaborn).

**4. Pré-processamento**

**4.1 Tratamento de Missing Values**

Foram aplicadas as seguintes estratégias:

- Variáveis numéricas: imputação pela mediana

- Variáveis categóricas: imputação pelo modo

- Verificação final: nenhum missing value restante após o processo

**4.2 Tratamento de Outliers**

- Boxplots indicaram outliers leves → não foram removidos, pois representavam comportamento real.

**4.3 Feature Engineering**

- Transformação de variáveis categóricas via One-Hot Encoding

- Padronização de variáveis numéricas via StandardScaler

- Criação automática de pipeline com ColumnTransformer

**4.4 Divisão dos Dados**

- 80% treino

- 20% teste

- Treino ainda passou por validação cruzada (CV)

**5. Modelagem**

**5.1 Modelos Testados**

- Regressão Linear

- Random Forest Regressor

- Gradient Boosting

- XGBoost Regressor

**5.2 Métricas Utilizadas**

- MSE

- RMSE

- MAE

- R²

**5.3 Comparação de Desempenho (antes da otimização)**

Random Forest e Gradient Boosting apresentaram melhor desempenho inicial.

**5.4 Escolha do Modelo Final**

O modelo escolhido para otimização foi o Random Forest, por apresentar o melhor equilíbrio entre desempenho e interpretabilidade.

**6. Otimização do Modelo**

**6.1 Método**

- RandomizedSearchCV

- Validação cruzada: 5 folds

**6.2 Hiperparâmetros Otimizados**

- n_estimators

- max_depth

- min_samples_split

- min_samples_leaf

- max_features

**6.3 Resultados Finais**

**Validação (melhor modelo)**

- MSE: 0.0365

- RMSE: 0.1910

- MAE: 0.1003

- R²: 0.9585

- CV RMSE: 0.2258

**Teste (desempenho final)**

- MSE: 0.0361

- RMSE: 0.1899

- MAE: 0.0987

- R²: 0.9645

O modelo generaliza extremamente bem, com estabilidade entre validação e teste.

**7. Conclusões**

**7.1 Resultados Alcançados**

- O modelo explica 96,45% da variância da nota final.

- A margem média de erro (MAE = 0.098) indica alta precisão.

- A otimização realmente melhorou o modelo base.

**7.2 Limitações**

- Alguns atributos como renda_familiar são categóricos simplificados e poderiam ter maior granularidade.

- O dataset não inclui fatores psicológicos, motivacionais ou qualidade do ensino.
